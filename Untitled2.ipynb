```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
url = 'https://data.cityofnewyork.us/api/views/6fi9-q3ta/rows.csv?accessType=DOWNLOAD'
df = pd.read_csv(url, sep=',')

# Convert date
df["hour_beginning"] = pd.to_datetime(df["hour_beginning"])
df["DayOfWeek"] = df["hour_beginning"].dt.day_name()

# Filter weekdays
weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
df_weekdays = df[df["DayOfWeek"].isin(weekdays)]

# Calculate mean pedestrian counts per weekday
weekday_means = df_weekdays.groupby("DayOfWeek")["Pedestrians"].mean().reindex(weekdays)

# Plot
plt.figure(figsize=(8,4))
plt.plot(weekday_means.index, weekday_means.values)
plt.xlabel("Day of Week")
plt.ylabel("Avg Pedestrian Count")
plt.title("Brooklyn Bridge Pedestrian Counts (Weekdays Only)")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
url = 'https://data.cityofnewyork.us/api/views/6fi9-q3ta/rows.csv?accessType=DOWNLOAD'
df = pd.read_csv(url, sep=',')

# Convert date & filter year 2019
df["hour_beginning"] = pd.to_datetime(df["hour_beginning"])
df_2019 = df[df["hour_beginning"].dt.year == 2019]

# Sort by weather summary
df_sorted = df_2019.sort_values("weather_summary")

# Select numeric weather columns
numeric_cols = ["temperature", "precipitation", "Pedestrians"]
corr = df_sorted[numeric_cols].corr()

# Plot heatmap
plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix: Weather vs Pedestrian Count (2019)")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
url = 'https://data.cityofnewyork.us/api/views/6fi9-q3ta/rows.csv?accessType=DOWNLOAD'
df = pd.read_csv(url, sep=',')

# Convert datetime columns
df["hour_beginning"] = pd.to_datetime(df["hour_beginning"])
df["Hour"] = df["hour_beginning"].dt.hour

# Custom time-of-day function
def categorize_time_of_day(hour):
    if 6 <= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 17:
        return "Afternoon"
    elif 18 <= hour <= 21:
        return "Evening"
    else:
        return "Night"

# Apply function
df["TimeOfDay"] = df["Hour"].apply(categorize_time_of_day)

# Group analysis
time_means = df.groupby("TimeOfDay")["Pedestrians"].mean().reindex(["Night","Morning","Afternoon","Evening"])

# Plot
plt.figure(figsize=(6,4))
plt.bar(time_means.index, time_means.values)
plt.xlabel("Time of Day")
plt.ylabel("Avg Pedestrian Count")
plt.title("Pedestrian Activity by Time of Day")
plt.show()
```
